```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.normal(loc=50, scale=10, size=200)
# sample data

plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.title("Histogram of Data Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(data)
plt.title("Boxplot Example")
plt.ylabel("Values")
plt.show()
```

```python
age = [22, 25, 30, 35, 40, 45, 50]
income = [25000, 30000, 40000, 50000, 60000, 65000, 70000]

plt.scatter(age, income, color='green')
plt.title("Scatter Plot: Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    'Age': age,
    'Income': income,
    'Experience': [1, 3, 7, 10, 15, 20, 25]
})

corr = df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

num_lines = 50
num_turns = 10
num_points = 1000000

fig, ax = plt.subplots(figsize=(6, 6))
theta = np.linspace(0, num_turns * 2 * np.pi, num_points)
r = np.linspace(0, 1, num_points)

x = r * np.cos(theta)
y = r * np.sin(theta)
ax.plot(x, y, color='black')

for i in range(num_lines):
    angle = 2 * np.pi * i / num_lines
    x_line = [0, np.cos(angle)]
    y_line = [0, np.sin(angle)]
    ax.plot(x_line, y_line, color='black', linewidth=0.8)

ax.axis('off')
plt.show()
```
