```python
def find_max(max_arr):
    max_val = max_arr[0]
    for i in range(1,len(max_arr)):
        if max_arr[i] > max_val:
            max_val = max_arr[i]
    return max_val


max_arr = [33,344,88]
print(find_max(max_arr))
```
