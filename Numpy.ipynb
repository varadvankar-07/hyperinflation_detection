```python
A = int(input("Enter number A: "))
B = int(input("Enter number B: "))
C = int(input("Enter number C: "))

D = B * B - 4 * A * C

print("Discriminant =", D)

if D > 0:
    x1 = (-B + D**0.5) / (2 * A)
    x2 = (-B - D**0.5) / (2 * A)

    print("x1 =", x1)
    print("x2 =", x2)
elif D == 0:
    x = -B / (2 * A)
    print("x =", x)
else:
    print("no real solution")
```
