```python
from Crypto.PublicKey import RSA
from Crypto.Signature import PKCS1_v1_5
from Crypto.Hash import SHA256

class Client:
    def __init__(self):
        self.key = RSA.generate(1024)
        self.public_key = self.key.publickey()

    def sign(self, msg):
        h = SHA256.new(msg.encode())
        return PKCS1_v1_5.new(self.key).sign(h)

    def verify(self, msg, sig):
        h = SHA256.new(msg.encode())
        return PKCS1_v1_5.new(self.public_key).verify(h, sig)


# Test
c = Client()
msg = "Hello Virat"

sig = c.sign(msg)
print("Valid?", c.verify(msg, sig))
print("Tampered?", c.verify("fake", sig))
```
