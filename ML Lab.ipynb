```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
pd.set_option('display.expand_frame_repr', False) # Disable wrapping to multiple rows

# Create a sample customer transactions dataset
data = {
    'CustomerID': [101, 102, 103, 104, 105, 106, 107],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None, 'Eve', 'Frank'],
    'Age': [25, 35, 45, np.nan, 50, 28, 22],
    'Gender': ['Female', 'Male', 'Male', 'Male', 'Female', 'Female', 'Male'],
    'Annual_Income': [50000, 60000, 45000, 70000, None, 30000, 40000],
    'Spending_Score': [80, 60, 40, 30, 50, 90, np.nan],
    'Membership': ['Gold', 'Silver', 'Gold', 'Bronze', 'Silver', 'Bronze', 'Gold'],

}

df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
```

```python
df["Name"] = df["Name"].fillna("Unknown")
print(df)
```

```python
# Impute missing 'Age' and 'Annual_Income' with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Annual_Income'] = df['Annual_Income'].fillna(df['Annual_Income'].mean())
print(df)
```

```python
imputer = SimpleImputer(strategy='median')
df[['Spending_Score']] = imputer.fit_transform(df[['Spending_Score']])
print(df)
```

```python
label_encoder_Gender = LabelEncoder()
df['Gender']=label_encoder_Gender.fit_transform(df['Gender'])
print(df)
```

```python
label_encoder_membership = LabelEncoder()
df['Membership'] = label_encoder_membership.fit_transform(df['Membership'])
print(df)
```

```python
#StandardScaler for 'Spending_Score'
scale = StandardScaler()
df[['Spending_Score']] = scale.fit_transform(df[['Spending_Score']])
print(df)
```

```python
#Standardscaler for 'Annual_income'
scale_Annual_income = StandardScaler()
df[['Annual_Income']] = scale_Annual_income.fit_transform(df[['Annual_Income']])
print(df)
```
